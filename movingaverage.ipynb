```python
import numpy as npimport matplotlib.pyplot as plt
```

```python
fs = 100t = np.linspace(0, 2, fs*2, endpoint=False)clean_signal = np.sin(2 * np.pi * 5 * t)noise = 0.5 * np.random.normal(size=t.shape)noisy_signal = clean_signal + noise
```

```python
def moving_average(signal, window_size):    window = np.ones(window_size) / window_size    return np.convolve(signal, window, mode='same')
```

```python
windows = [4, 9, 20]filtered = {w: moving_average(noisy_signal, w) for w in windows}
```

```python
plt.figure(figsize=(10,6))plt.plot(t, noisy_signal, label='Noisy Input', color='orange')for w in windows:    plt.plot(t, filtered[w], label=f'Window = {w}')plt.legend()plt.xlabel('Time [s]')plt.ylabel('Amplitude')plt.title('Moving Average Filter - Different Window Sizes')plt.grid(True)plt.show()
```

```python
N = len(noisy_signal)freqs = np.fft.rfftfreq(N, d=1/fs)fft_noisy = np.abs(np.fft.rfft(noisy_signal)) / Nfft_filtered = {w: np.abs(np.fft.rfft(filtered[w])) / N for w in windows}
```

```python
plt.figure(figsize=(10,6))plt.plot(freqs, fft_noisy, label='Noisy Input', color='orange')for w in windows:    plt.plot(freqs, fft_filtered[w], label=f'Window = {w}')plt.xlabel('Frequency [Hz]')plt.ylabel('Magnitude')plt.title('FFT - Moving Average Filter')plt.legend()plt.grid(True)plt.show()
```
