```python
import pandas as pdimport numpy as npfrom scipy.signal import medfiltimport matplotlib.pyplot as plt
```

```python
data = pd.read_csv("measured_raman_data_with_headers.csv")wavenumber = data["Wavenumber_cm-1"]intensity = data["Measured_Intensity"]
```

```python
plt.figure(figsize=(10,5))plt.plot(wavenumber, intensity, color="gray")plt.xlabel("Raman Shift (cm-1)")plt.ylabel("Intensity (Counts)")plt.title("Measured Raman Spectrum")plt.show()
```

```python
filtered = medfilt(intensity, kernel_size=5)
```

```python
plt.figure(figsize=(10,5))plt.plot(wavenumber, intensity, color="gray", label="Measured (With Cosmic Spikes)")plt.plot(wavenumber, filtered, color="red", label="Filtered (Moving Median)")plt.xlabel("Raman Shift (cm-1)")plt.ylabel("Intensity (Counts)")plt.title("Raman Spectrum Processing: Removing Cosmic Ray Spikes")plt.legend()plt.show()
```
