```python
import pandas as pdimport numpy as npimport matplotlib.pyplot as plt
```

```python
signal = pd.read_csv("noisy_signal_1.csv", header=None).iloc[0].dropna().reset_index(drop=True)t = np.linspace(0, 5, len(signal))
```

```python
plt.figure(figsize=(10,5))plt.plot(t, signal)plt.xlabel("Time (Seconds)")plt.ylabel("Sensor Value")plt.title("Noisy Measurement")plt.show()
```

```python
moving_avg = signal.rolling(window=9, center=True, min_periods=1).mean()
```

```python
moving_median = signal.rolling(window=9, center=True, min_periods=1).median()
```

```python
hybrid = moving_median.rolling(window=9, center=True, min_periods=1).mean()
```

```python
plt.figure(figsize=(10,5))plt.plot(t, signal, color="gray", label="Noisy Measurement")plt.plot(t, moving_avg, color="blue", label="Moving Average")plt.plot(t, moving_median, color="red", label="Moving Median")plt.plot(t, hybrid, color="green", label="Hybrid (Median + Avg)")plt.xlabel("Time (Seconds)")plt.ylabel("Sensor Value")plt.legend()plt.show()
```

```python
windows = [4, 9, 20]moving_avgs = {w: signal.rolling(window=w, center=True, min_periods=1).mean() for w in windows}
```

```python
plt.figure(figsize=(10,5))plt.plot(t, signal, color="gray", label="Noisy Measurement")for w in windows:    plt.plot(t, moving_avgs[w], label=f"Window = {w}")plt.xlabel("Time (Seconds)")plt.ylabel("Sensor Value")plt.title("Moving Average Filter - Different Window Sizes")plt.legend()plt.show()
```
