```python
import pandas as pdimport numpy as npfrom scipy.signal import medfiltimport matplotlib.pyplot as plt
```

```python
data = pd.read_csv("raw_raman_data.csv")wavenumber = data["Wavenumber_cm-1"]intensity = data["Raw_Intensity"]
```

```python
plt.figure(figsize=(10,5))plt.plot(wavenumber, intensity, color="gray")plt.xlabel("Wavenumber (cm-1)")plt.ylabel("Intensity")plt.title("Raw Raman Signal")plt.show()
```

```python
baseline = medfilt(intensity, kernel_size=201)
```

```python
corrected = intensity - baseline
```

```python
plt.figure(figsize=(10,5))plt.plot(wavenumber, intensity, color="gray", label="Raw Measured")plt.plot(wavenumber, baseline, color="red", label="Estimated Baseline (Wide Window)")plt.plot(wavenumber, corrected, color="blue", label="Baseline Corrected Signal")plt.xlabel("Wavenumber (cm-1)")plt.ylabel("Intensity")plt.title("Raw Raman Spectrum: Baseline Removal")plt.legend()plt.show()
```
